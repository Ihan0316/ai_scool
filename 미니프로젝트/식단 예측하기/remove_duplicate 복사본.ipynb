```python
import os
import hashlib
from PIL import Image

def calculate_image_hash(file_path):
    with Image.open(file_path) as img:
        img_hash = hashlib.md5(img.tobytes()).hexdigest()
    return img_hash

def remove_duplicate_images(directory):
    hash_dict = {}
    duplicates_removed = 0

    for root, _, files in os.walk(directory):
        for filename in files:
            if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp')):
                file_path = os.path.join(root, filename)
                try:
                    img_hash = calculate_image_hash(file_path)
                    if img_hash in hash_dict:
                        os.remove(file_path)
                        duplicates_removed += 1
                        print(f"중복 이미지 제거: {file_path}")
                    else:
                        hash_dict[img_hash] = file_path
                except Exception as e:
                    print(f"오류 발생: {file_path} - {str(e)}")

    print(f"총 {duplicates_removed}개의 중복 이미지가 제거되었습니다.")

# 사용 예시
directory_path = ('img')
remove_duplicate_images(directory_path)
```
