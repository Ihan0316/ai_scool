```python
import os
import re
import requests
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

url = 'https://pf.kakao.com/_Ebxors/posts'

driver = webdriver.Chrome()  # chromedriver 경로 설정 필요
driver.get(url)
time.sleep(3)  # 페이지 초기 로딩 대기

# 이미지 저장 폴더 생성
os.makedirs("img", exist_ok=True)

prev_img_count = 0
scroll_attempt = 0
max_scroll_attempts = 10  # 최대 스크롤 시도 횟수 설정

while True:
    # 모든 div.wrap_fit_thumb 요소 선택
    thumb_divs = driver.find_elements(By.CLASS_NAME, "wrap_fit_thumb")
    current_img_count = len(thumb_divs)

    if current_img_count == 0:
        # 이미지가 아직 안 뜬 경우 스크롤
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)
        scroll_attempt += 1
        if scroll_attempt >= max_scroll_attempts:
            print("더 이상 스크롤할 이미지가 없음")
            break
        continue

    # 이미지가 새로 생겼으면 새로 다운받기
    if current_img_count > prev_img_count:
        for idx in range(prev_img_count, current_img_count):
            div = thumb_divs[idx]
            style = div.get_attribute("style")
            match = re.search(r'url\([\'"]?(.*?)[\'"]?\)', style)
            if match:
                img_url = match.group(1)
                try:
                    res = requests.get(img_url)
                    file_path = f"img/thumb_{idx}.jpg"
                    with open(file_path, "wb") as f:
                        f.write(res.content)
                    print(f"저장 완료: {file_path}")
                except Exception as e:
                    print(f"다운 실패: {img_url} | 에러: {e}")
        prev_img_count = current_img_count
        scroll_attempt = 0
    else:
        # 새로운 이미지가 없다면 다시 스크롤
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)
        scroll_attempt += 1
        if scroll_attempt >= max_scroll_attempts:
            print("더 이상 스크롤할 이미지가 없음")
            break

driver.quit()
```
