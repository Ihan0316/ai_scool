```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

src = cv2.imread('image/scenery.jpg', cv2.IMREAD_COLOR_RGB)

aff = np.array([[1, 0, 100], [0,1,100]], dtype=np.float32)
dst = cv2.warpAffine(src, aff, (0,0))

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(src)
plt.title('Source Image')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Affine Transformation')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

src = cv2.imread('image/scenery.jpg', cv2.IMREAD_COLOR_RGB)

aff = np.array([[1, 0.5, 0], [0,1,0]], dtype=np.float32)

h,w = src.shape[:2]

dst = cv2.warpAffine(src, aff, (w+int(h*0.5),h))

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(src)
plt.title('Source Image')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Affine Transformation')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

src = cv2.imread('image/candies.png', cv2.IMREAD_COLOR_RGB)

dst1 = cv2.resize(src, (0,0), fx=4, fy=4, interpolation=cv2.INTER_NEAREST)
dst2 = cv2.resize(src, (1920, 1280))
dst3 = cv2.resize(src, (1920, 1280), interpolation=cv2.INTER_CUBIC)
dst4 = cv2.resize(src, (1920, 1280), interpolation=cv2.INTER_LANCZOS4)

plt.figure(figsize=(20,10))

plt.subplot(151)
plt.imshow(src)
plt.title('Source Image')
plt.axis('off')

plt.subplot(152)
plt.imshow(dst1[500:900, 400:800])
plt.title('Nearest Neighbor')
plt.axis('off')

plt.subplot(153)
plt.imshow(dst2[500:900, 400:800])
plt.title('Bilinear')
plt.axis('off')

plt.subplot(154)
plt.imshow(dst3[500:900, 400:800])
plt.title('Cubic')
plt.axis('off')

plt.subplot(155)
plt.imshow(dst4[500:900, 400:800])
plt.title('Lanczos')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import math
from matplotlib import pyplot as plt

src = cv2.imread('image/scenery.jpg', cv2.IMREAD_COLOR_RGB)

rad = 20 * math.pi / 180
aff = np.array([[math.cos(rad), math.sin(rad), 0], [-math.sin(rad), math.cos(rad), 0]], dtype=np.float32)
dst = cv2.warpAffine(src, aff, (0,0))

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(src)
plt.title('Source Image')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Affine Transformation')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import math
from matplotlib import pyplot as plt

src = cv2.imread('image/scenery.jpg', cv2.IMREAD_COLOR_RGB)
cp = (src.shape[1]/2, src.shape[0]/2)
rot = cv2.getRotationMatrix2D(cp, 20, 0.5)
dst = cv2.warpAffine(src, rot, (0,0))

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(src)
plt.title('Source Image')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Affine Transformation')
plt.axis('off')

plt.show()
```
