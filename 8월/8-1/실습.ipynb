```python
import cv2
import numpy as np
import math
from matplotlib import pyplot as plt

# 문제 1.
# 이미지 'sample.jpg'를 불러와서 5x5 커널의 평균값 필터링을 적용하고, 원본 이미지와 결과 이미지를 나란히 출력하세요.
img = cv2.imread('../../sample/pier39.JPG', cv2.IMREAD_COLOR_RGB)

kernel = np.ones((5,5),np.float32)/25
dst = cv2.filter2D(img,-1,kernel)

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(img)
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Averaging')
plt.axis('off')

plt.show()
```

```python
# 문제 2.
# 이미지에 대해 5x5 커널과 sigmaX=1.5를 갖는 가우시안 필터를 적용하세요.

dst = cv2.GaussianBlur(img, (5,5), 1.5)

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(img)
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Gaussian Blur')
plt.axis('off')

plt.show()
```

```python
# 문제 3.
# 노이즈가 섞인 이미지에 대해 커널 크기 5의 미디언 필터를 적용하세요.

dst = cv2.medianBlur(img, 5)

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(img)
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Median Blur')
plt.axis('off')

plt.show()
```

```python
# 문제 4.
# 이미지를 흐리게 만든 뒤 원본 이미지와 가중합을 계산하여 샤프닝 효과를 구현하세요.

blur = cv2.GaussianBlur(img,(5,5),0)
sharp = cv2.addWeighted(img,1.5,blur,-0.5,0)

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(blur)
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(sharp)
plt.title('Sharp')
plt.axis('off')

plt.show()
```

```python
# 문제 5.
# 양방향 필터를 사용하여 이미지에서 노이즈를 제거하되 에지는 보존하세요.

dst = cv2.bilateralFilter(img, -1, 10, 5)

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(img)
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Bilateral Filter')
plt.axis('off')

plt.show()
```

```python
# 문제 6.
# 이미지 'sample.jpg'를 x축으로 50픽셀, y축으로 30픽셀만큼 이동시키는 코드를 작성하세요. 이동을 위한 어파인 행렬을 생성하고, cv2.warpAffine()을 이용해 결과를 시각화하세요.

aff = np.array([[1,0,50],[0,1,30]], dtype=np.float32)
dst = cv2.warpAffine(img, aff, (0,0))

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(img)
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Warp Affine')
plt.axis('off')

plt.show()
```

```python
# 문제 7.
# 이미지를 가로 세로 2배로 확대하고, 다시 0.5배로 축소하는 코드를 작성하세요. interpolation 옵션을 다르게 설정하고, 그 차이를 비교해보세요.

dst1 = cv2.resize(img, (0,0), fx=2.0, fy=2.0, interpolation=cv2.INTER_LANCZOS4)
dst2 = cv2.resize(dst1, (0,0), fx=0.5, fy=0.5, interpolation=cv2.INTER_LANCZOS4)
plt.figure(figsize=(20,10))

plt.subplot(131)
plt.imshow(img[200:600, 400:800])
plt.title('Original')
plt.axis('off')

plt.subplot(132)
plt.imshow(dst1[400:800, 800:1200])
plt.title('2X zoom')
plt.axis('off')

plt.subplot(133)
plt.imshow(dst2[200:600, 400:800])
plt.title('0.5X zoom')
plt.axis('off')

plt.show()
```

```python
# 문제 8.
# 이미지 중심을 기준으로 반시계 방향으로 45도 회전하는 코드를 작성하세요. 회전 행렬을 만들고, 회전 결과를 출력하세요.

rad = 20*math.pi/180
aff = np.array([[math.cos(rad), math.sin(rad), 0], [-math.sin(rad), math.cos(rad), 0]], dtype=np.float32)
dst = cv2.warpAffine(img, aff, (0,0))

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(img)
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Rotation')
plt.axis('off')

plt.show()
```
