```python
import cv2, sys
from matplotlib import pyplot as plt

src = cv2.imread('../../sample/building.png', cv2.IMREAD_GRAYSCALE)

fast = cv2.FastFeatureDetector_create(60)
keypoint = fast.detect(src)

dst = cv2.cvtColor(src, cv2.COLOR_GRAY2BGR)

for kp in keypoint:
    pt = (int(kp.pt[0]), int(kp.pt[1]))
    cv2.circle(dst, pt, 5, (0, 0, 255), 2)

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(src)
plt.title('Source Image')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst)
plt.title('Feature Detection')
plt.axis('off')

plt.show()
```

```python
import cv2, sys

src1 = cv2.imread('../../sample/box.png', cv2.IMREAD_GRAYSCALE)
src2 = cv2.imread('../../sample/box_in_scene.png', cv2.IMREAD_GRAYSCALE)

feature = cv2.KAZE_create()

kp1 = feature.detect(src1)
kp2 = feature.detect(src2)

print('# of kp1:', len(kp1))
print('# of kp2:', len(kp2))

dst1 = cv2.drawKeypoints(src1, kp1, None, flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)
dst2 = cv2.drawKeypoints(src2, kp2, None, flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

plt.figure(figsize=(20,10))

plt.subplot(121)
plt.imshow(dst1)
plt.title('Source Image')
plt.axis('off')

plt.subplot(122)
plt.imshow(dst2)
plt.title('Feature Detection')
plt.axis('off')

plt.show()
```
