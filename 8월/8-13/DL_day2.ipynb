```python
#0. 작업 준비
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
#데이터 분할
from sklearn.model_selection import train_test_split
#모델 모듈(딥러닝)
from keras.models import Sequential
from keras.layers import Input,Dense,Conv2D,MaxPooling2D,Dropout,SimpleRNN,LSTM,GRU,BatchNormalization
from keras.optimizers import Adam,RMSprop,SGD,Adagrad
from keras.losses import binary_crossentropy,categorical_crossentropy,sparse_categorical_crossentropy,mean_squared_error
#모델 학습
from keras.callbacks import EarlyStopping,ModelCheckpoint
#모델 로드
from keras.models import load_model
#성능검토
from sklearn.metrics import r2_score,classification_report,f1_score,mean_squared_error
from tensorflow import truediv
```

```python
from keras.datasets import fashion_mnist
(x_data, y_data),(tt_x, tt_y) = fashion_mnist.load_data()
```

```python
tr_x, val_x, tr_y, val_y = train_test_split(x_data, y_data, test_size=0.8, stratify=y_data)
```

```python
np.unique(tr_x/255) # sigmoid는 왠만하면 스케일링 해주기
```

```python
tr_x.shape, tr_y.shape,tt_x.shape, tt_y.shape, val_x.shape, val_y.shape
```

```python
e_tr_x = tr_x/255
```

```python
e_tt_x=tt_x/255
e_val_x=val_x/255
```

```python
np.unique(tr_y)
```

```python
from keras.layers import Flatten
m = Sequential()
m.add(Input(shape=tr_x.shape[1:]))
m.add(Flatten())
m.add(Dense(100, activation='sigmoid')) # 은닉
m.add(Dense(100, activation='sigmoid')) # 은닉
m.add(Dense(10, activation='sigmoid'))  # 출력

m.summary()
```

```python
m.compile(loss='sparse_categorical_crossentropy', optimizer='sgd', metrics=['acc'])
```

```python
es = EarlyStopping(patience=5, verbose=1, mode='auto', restore_best_weights=True)
ck = ModelCheckpoint('best_m.keras', save_best_only=True, verbose=1)
```

```python
hr = m.fit(e_tr_x, tr_y, validation_data=(e_val_x, val_y), epochs=100, callbacks=[es,ck])
```

```python
plt.plot(hr.history['loss'])
plt.plot(hr.history['val_loss'])
```

```python
#0. 작업 준비
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
#데이터 분할
from sklearn.model_selection import train_test_split
#모델 모듈(딥러닝)
from keras.models import Sequential
from keras.layers import Input,Dense,Conv2D,MaxPooling2D,Dropout,SimpleRNN,LSTM,GRU,BatchNormalization, Flatten
from keras.optimizers import Adam,RMSprop,SGD,Adagrad
from keras.losses import binary_crossentropy,categorical_crossentropy,sparse_categorical_crossentropy,mean_squared_error
#모델 학습
from keras.callbacks import EarlyStopping,ModelCheckpoint
#모델 로드
from keras.models import load_model
#성능검토
from sklearn.metrics import r2_score,classification_report,f1_score,mean_squared_error
from tensorflow import truediv
```

```python
from keras.datasets import cifar10
```

```python
(x_data, y_data), (tt_x, tt_y) = cifar10.load_data()
```

```python
tr_x,val_x,tr_y,val_y = train_test_split(x_data, y_data, train_size=0.8, stratify=y_data)
```

```python
tr_x.shape, tr_y.shape,tt_x.shape, tt_y.shape, val_x.shape, val_y.shape
```

```python
tr_y
```

```python
m = Sequential()
m.add(Input(shape=tr_x.shape[1:]))
m.add(Flatten())
m.add(Dense(100, activation='relu'))
m.add(Dense(100, activation='relu'))
m.add(Dense(100, activation='relu'))
m.add(Dense(10, activation='softmax'))
m.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['acc'])
```

```python
m.summary()
```

```python
es = EarlyStopping(patience=10, restore_best_weights=True)
ck = ModelCheckpoint('best_m.keras', save_best_only=True)
```

```python
hy = m.fit(tr_x, tr_y, validation_data=(val_x, val_y), epochs=100, callbacks=[es,ck])
```

```python
plt.plot(hy.history['loss'])
plt.plot(hy.history['val_loss'])
```

```python
m = Sequential()
m.add(Input(shape=tr_x.shape[1:]))
m.add(Conv2D(10, 3, 1, activation='relu', padding='same')) # 32,32,10
m.add(MaxPooling2D(2)) # 16,16,10
m.add(Conv2D(10, 3, 1, activation='relu', padding='same')) # 16,16,10
m.add(MaxPooling2D(2))# 8,8,10
m.add(Conv2D(10, 3, 1, activation='relu', padding='same')) # 8,8,10
m.add(MaxPooling2D(2))# 4,4,10
m.add(Conv2D(10, 3, 1, activation='relu', padding='same')) # 4,4,10
m.add(MaxPooling2D(2))# 2,2,10
m.add(Conv2D(10, 3, 1, activation='relu', padding='same')) # 2,2,10
m.add(MaxPooling2D(2))# 1,1,10
m.add(Flatten())
m.add(Dense(10, activation='softmax'))

m.summary()
```
