```python
text="A barber is a person. a barber is good person. a barber is huge person. he Knew A Secret! The Secret He Kept is huge secret. Huge secret. His barber kept his word. a barber kept his word. His barber kept his secret. But keeping and keeping such a huge secret to himself was driving the barber crazy. the barber went up a huge mountain."
문장_list=sent_tokenize(text)
문장_list
```

```python
vocab={}
전처리된_문장집합=[]
불용어_집합=set(stopwords.words('english'))

for 문장 in 문장_list:
    토큰화된_문장=word_tokenize(문장)

    정리_문장=[]
    for 단어 in 토큰화된_문장:
        소문자화된_단어=단어.lower()
        if 소문자화된_단어 not in 불용어_집합:
            if len(소문자화된_단어)>2:
                정리_문장.append(소문자화된_단어)
                if 소문자화된_단어 not in vocab:
                    vocab[소문자화된_단어]=0
                vocab[소문자화된_단어]+=1
    전처리된_문장집합.append(정리_문장)
전처리된_문장집합
```

```python
vocab
```

```python
정렬된_단어집합=sorted(vocab.items(),key=lambda x:x[1],reverse=True)
```

```python
단어_idx={}
i=0
for (단어,빈도수) in 정렬된_단어집합:
    if 빈도수 >1:
        i+=1
        단어_idx[단어]=i
단어_idx
```

```python
vocab_size=5
제거될_단어들=[단어 for 단어,idx in 단어_idx.items() if idx>=vocab_size+1]
for 제거용단어 in 제거될_단어들:
    del 단어_idx[제거용단어]
단어_idx    
```

```python
단어_idx['OOV']=len(단어_idx)+1
단어_idx
```

```python
#정수 인코딩
인코딩된_문장집합=[]
for 문장 in 전처리된_문장집합:
    인코딩된_문장=[]
    for 단어 in 문장:
        try:
            인코딩된_문장.append(단어_idx[단어])
        except KeyError:
            인코딩된_문장.append(단어_idx['OOV'])
    인코딩된_문장집합.append(인코딩된_문장)
인코딩된_문장집합  
```

```python
#0. 작업 준비
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from tensorflow.keras.preprocessing.text import Tokenizer
from keras.utils import pad_sequences
from keras.utils import to_categorical
from sklearn.preprocessing import LabelEncoder
from nltk.corpus import stopwords
import re

from sklearn.model_selection import train_test_split

from keras.models import load_model
from keras.callbacks import EarlyStopping,ModelCheckpoint
```

```python
def 전처리기(문장,불용어_사용유뮤=True):
    문장정리_1차=re.sub('[^a-zA-Z0-9]',' ',문장)
    소문자화_문장=문장정리_1차.lower().split()
    if 불용어_사용유뮤:
        불용어=set(stopwords.words('english'))
        정리된_문장=[단어 for 단어 in 소문자화_문장 if 단어 not in 불용어]
        최종_문장 = ' '.join(정리된_문장)
    else:
        최종_문장 = ' '.join(소문자화_문장)
    return 최종_문장
```

```python
df['X']=df.v2.apply(lambda x: 전처리기(x))
```

```python
df=df.drop_duplicates(subset=['X'])
df.loc[:,'X']=df['X'].replace('[^a-zA-Z0-9 ]','',regex=True)
df.loc[:,'X']=df['X'].replace('^ +','',regex=True)
df.loc[:,'X']=df['X'].replace('',np.nan)
df=df.dropna(how='any')
```
